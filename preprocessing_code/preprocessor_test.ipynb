```python
import sys
from pathlib import Path

# Make sure we can import from ../src
sys.path.append(str(Path('..') / 'src'))

from audio_preprocessor import AudioPreprocessor
from IPython.display import Audio, display

# Pick an example audio file (WAV/FLAC recommended).
candidates = []
for ext in ('.wav', '.flac', '.ogg'):
    candidates.extend(list(Path('..').rglob(f'*{ext}')))
if not candidates:
    raise FileNotFoundError('No .wav/.flac/.ogg found under repo. Set a path manually.')
input_audio = candidates[0]
print('Using:', input_audio)

preprocessor = AudioPreprocessor()

original_audio, sr = preprocessor.load_audio(input_audio)
print('original audio')
display(Audio(original_audio, rate=sr))

processed_audio, sr = preprocessor.preprocess(input_audio)
print('processed audio')
display(Audio(processed_audio, rate=sr))
```
